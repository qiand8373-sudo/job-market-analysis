# 01 数据概览 — 第一次接触数据

> 今天要学的知识点：pandas 读取 CSV、看前几行、看规模、看类型、看统计。
>
> 数据集：前程无忧数据分析相关岗位 6555 条（2019 年）。

**运行方式**：从第一个 cell 开始，逐个选中按 `Shift+Enter` 运行，**先猜输出再按**，猜完对照实际输出。

```python
# 知识点：import 是「导入库」的意思，pandas 是数据分析最核心的库
# 「as pd」是给它起个短名字，以后写 pd.xxx 就是调用它（行业约定俗成）
import pandas as pd

# 知识点：pd.read_csv('路径') 读取 CSV 文件，返回一个 DataFrame（表格）
# DataFrame = 数据岗每天打交道的对象：有行有列，像 Excel 工作表
# 「df」是变量名，把读到的表格存进去
df = pd.read_csv('../data/raw/item_detail.csv')

# 知识点：df.head() 显示前 5 行，快速看看数据长什么样
df.head()
```

```python
# 知识点：df.shape 返回 (行数, 列数)，不用打印就能看到数据规模
# 猜一猜：这个数据集有多少行？多少列？
df.shape
```

```python
# 知识点：df.info() 显示每列的信息：
#  - 列名
#  - 非空数量（如果和总行数不一样，说明这列有缺失值）
#  - Dtype（数据类型）：object=文本，int64/float64=数字
# 仔细看哪几列是文本、哪几列是数字
df.info()
```

```python
# 知识点：df.describe() 只统计数值列，输出描述性统计：
#  count=非空个数，mean=平均值，std=标准差（波动大小）
#  min=最小值，25%/50%/75%=分位数，max=最大值
# 只看 salary(k/m) 这一列：最低多少？平均多少？最高多少？
df.describe()
```

## 今天的 4 个知识点

| 代码 | 作用 |
|------|------|
| `pd.read_csv('路径')` | 读 CSV 成 DataFrame |
| `df.head()` | 看前 5 行 |
| `df.shape` | 看行数×列数 |
| `df.info()` | 看每列类型和缺失情况 |
| `df.describe()` | 数值列统计 |
