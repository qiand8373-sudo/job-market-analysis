# 02 最小闭环分析 — 基本统计 + 第一张图

> 今天要学的知识点：value_counts（计数）、groupby（分组聚合）、matplotlib 画柱状图。
>
> **最小闭环**的意思：从原始数据出发，经过统计，输出一个「能说给别人听」的结论。

运行方式：逐个 cell 按 `Shift+Enter`，先猜输出再运行。

```python
import pandas as pd

df = pd.read_csv('../data/raw/item_detail.csv')

# 知识点：df['列名'] 取出某一列（叫 Series，只有一列的表）
# 知识点：.value_counts() 统计这一列每个值出现了多少次，自动从多到少排序
# 知识点：.head(10) 只要前 10 个
# 猜：哪个城市出现最多？
df['job_loc_city'].value_counts().head(10)
```

```python
# 知识点：df.groupby('列A') 按 A 列分组（把相同城市的行归到一组）
# 知识点：['salary(k/m)'].mean() 对每组求薪资平均值
# 知识点：.sort_values(ascending=False) 按值从大到小排序
df.groupby('job_loc_city')['salary(k/m)'].mean().sort_values(ascending=False).head(10)
```

```python
# ===== 人生第一张图 =====
# 知识点：import matplotlib.pyplot as plt 导入画图库
import matplotlib.pyplot as plt

# 知识点：这两行解决中文乱码（Windows 用 SimHei 黑体）
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 知识点：Series.plot(kind='bar') 直接画柱状图
counts = df['job_loc_city'].value_counts().head(10)
counts.plot(kind='bar', title='岗位数量最多的 10 个城市')
plt.show()
```

## 👀 观察：这张图好看吗？

**不好看，但这是真实的数据。**

成都 6283 条、第二名才 34 条——这个数据集是爬虫按「成都」搜索的，其他城市只是附带爬到的。

**这是数据分析的第一课**：真实数据永远不完美。分析师的职责不是把图变好看，而是**发现数据的问题**并判断它还能回答什么、不能回答什么。

→ 结论：这个数据集**不能**回答「哪个城市岗位最多」（数据有偏），但**能**回答学历分布、薪资分布、经验要求等不依赖城市的问题。

```python
# 换个没被爬虫污染的角度：学历要求分布
# 猜：要求最多的学历是什么？
edu_counts = df['edu'].value_counts()
edu_counts.plot(kind='bar', title='学历要求分布', color='steelblue')
plt.show()
```

```python
# 进阶一点：大专+本科占了多少比例？
# 知识点：算比例 = 每个值 / 总和
# 知识点：round(2) 保留两位小数
(df['edu'].value_counts() / len(df) * 100).round(2)
```
